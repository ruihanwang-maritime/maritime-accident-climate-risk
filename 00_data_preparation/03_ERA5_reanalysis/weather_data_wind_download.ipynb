```python
import os
DATA_ROOT = os.environ.get('WRH_DATA_ROOT', 'G:/WRH_data')  # local data folder; see README
import cdsapi
import os
from pathlib import Path

# ── target directory ──────────────────────────────────────────────────────────
SAVE_DIR = Path(rf"{DATA_ROOT}/AIS_processing/weather_data/wind")
SAVE_DIR.mkdir(parents=True, exist_ok=True)

dataset = "reanalysis-era5-single-levels"
client  = cdsapi.Client()

MONTHS = [f"{m:02d}" for m in range(1, 13)]
DAYS   = [f"{d:02d}" for d in range(1, 32)]
HOURS  = [f"{h:02d}:00" for h in range(24)]

# ── download by year (resume: skip existing)────────────────────────────────
for year in range(2016, 2018):
    target = SAVE_DIR / f"era5_wind10m_{year}.grib"

    if target.exists():
        print(f"[SKIP]  {year}  →  {target.name} already exists")
        continue

    print(f"[QUEUE] {year}  →  {target.name} ...")
    request = {
        "product_type": ["reanalysis"],
        "variable": [
            "10m_u_component_of_wind",
            "10m_v_component_of_wind",
        ],
        "year":  [str(year)],
        "month": MONTHS,
        "day":   DAYS,
        "time":  HOURS,
        "data_format":    "netcdf",
        "download_format": "unarchived",
    }

    client.retrieve(dataset, request).download(target=str(target))
    size_gb = target.stat().st_size / 1e9
    print(f"[DONE]  {year}  →  {target.name}  ({size_gb:.1f} GB)")

print("\nAll years complete.")
print(f"Total files: {len(list(SAVE_DIR.glob('*.grib')))}")
print(f"Total size : {sum(f.stat().st_size for f in SAVE_DIR.glob('*.grib'))/1e9:.1f} GB")
```
